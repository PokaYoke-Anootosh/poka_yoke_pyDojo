* GroupBy method in pandas:

```python
car_sales.groupby(["Make"]).mean(numeric_only=True)
```

So groupby detects all columns in the dataframe having a numeric value and returns the mean for all of those numeric columnns. Note that in order to make it auto-detect columns with numeric values only and skip columns with  alphabetic or other values, we need to specify the numeric_only=True. 

Similarly we can also mention the column names for the only columns we need instead of doing an aggregation (mean/average, sum, etc.) on all numeric columns as shown below.

```python
# Here I have used 2 columns to group the data by...we can use multiple columns to 
# group the data:
car_sales.groupby(["Make","Doors"])["Odometer (KM)"].mean()
```

```python
# Grouping all the White Toyotas in the car-sales data:

car_sales.groupby(["Make","Colour"]).mean(numeric_only=True).loc[("Toyota","White")]
```

```python
# OR a cleaner approach: 

car_sales.loc[(car_sales["Make"] == "Toyota") & (car_sales["Colour"] == "White")].mean(numeric_only=True)
```

* Converting the price column into integer before plotting it:

```python
# Identifying the data type of the price column 
car_sales["Price"].dtype
```

```python
# Since it returns StringDtype above meaning the values in the Price column 
# are strings and not numbers due to the dollar sign ($) in front of the numbers, 
# we need to convert the Dtype to interger or float. 

# To convert to float or floating point number -
car_sales["Price"] = car_sales["Price"].str.replace(r'[\$,]', '', regex=True).astype(float)

#Viewing the converted Column values - 
car_sales["Price"]
```

```python
# Plotting the values - 
car_sales["Price"].plot()
```

```python
# Reloading the table to run the Covert to interger code - 
car_sales_Reload = pd.read_csv("car-sales.csv")

# To convert to integer - 
car_sales_Reload["Price"] = car_sales_Reload["Price"].str.replace(r'[\$,.]', '', regex=True).astype(int)

#Viewing the converted Column values - 
car_sales_Reload["Price"]

```

```python
# Plotting the values - 
car_sales_Reload["Price"].plot()
```

```python
# Tracking Missing data in a dataset/dataframe - 

car_sales_w_missing_val = pd.read_csv("car-sales-missing-data.csv")

car_sales_w_missing_val
```

The above data has rows/records with missing/empty values that show up as 'NaN'.  

```python
# Let's replace the NaN/missing/blank values with something more useful like a 
# a marker that does not make our calculations go wrong, like may be a 0. 

# We use .fillna function to replace the NaN values in the odometer column 
# with a 0: 

car_sales_w_missing_val["Odometer"].fillna(0)
```

```python
# But these changes have yet been done to the actual car_sales_w_missing_val 
# dataframe/table. You'll see that when we load the table below. 

car_sales_w_missing_val
```

```python
# Now, to make these changes in the actual dataset permanently we will simply assign 
# the whole thing to the actual dataframe/table column like we did for str.lower(): 

car_sales_w_missing_val["Odometer"] = car_sales_w_missing_val["Odometer"].fillna(0)

car_sales_w_missing_val
```

```python
# Lets do the same for the price column too: 

car_sales_w_missing_val["Price"] = car_sales_w_missing_val["Price"].fillna(0) 

car_sales_w_missing_val
```

```python
# Dropping or Removing the NaN/blank values with .dropna function: 

car_sales_cleaned = car_sales_w_missing_val.dropna()

car_sales_cleaned
```

In the above dataframe (car_sales_cleaned) you will see that the rows with NaN values appearing under the Doors column are now gone. So any row with a NaN value has now been dropped using dropna function.
