```python
#but what if I changed the indexes (or index names) of each of the rows? 

My_Cars = car_sales.head(6)

My_Cars
```

```python
My_Cars.index = [0,2,4,5,4,6]

My_Cars
```

```python
#Returns 2 items with the same index 4 with .loc (after we changed the indexes)
My_Cars.loc[4] 
```

```python
#Now we use .iloc (for Positional Index location):

My_Cars.iloc[4]
```

```python
#.iloc does not return 2 items with same index 4, but only returns 
# 'Nissan' i.e. the value which is positioned 4th in the list of My_Cars 
# if we count from 0 to 6 (0, 1, 2, 3, 4, 5, 6 where 4 is the 5th index 
# position.
```

* .loc refers to the Index 
* .iloc refers to the Position 

* List Slicing with .iloc - 

```python
car_sales.iloc[:4]
```

```python
car_sales.iloc[1::4]
```
