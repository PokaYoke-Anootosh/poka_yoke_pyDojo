```python
import pandas as pd
```

* Series (1-dimensional):

```python
Car_make = pd.Series(["BMW", "Toyota", "Honda", "Tesla", "Mercedes"])
```

```python
Country = pd.Series(["Germany", "Japan", "Japan", "USA", "Germany"])
```

* DataFrames (2-dimensional):

```python
car_data = pd.DataFrame({
    "Make" : Car_make,
    "Country" : Country
})

car_data
```
